## Classification of Variables

```python
num_cols = ['Attendance_%', 'Study_Hours', 'Assignment_Marks', 'Internal_Marks', 'Final_Exam_Marks']
df_num = df[num_cols]
```

## Data Visualization

```python
plt.hist(df['Final_Exam_Marks'], bins=10, edgecolor='black')
plt.xlabel('Final Exam Marks')
plt.ylabel('Number of Students')
plt.title('Distribution of Final Exam Marks')
plt.show()
```

```python
plt.scatter(df['Study_Hours'], df['Final_Exam_Marks'])
plt.xlabel('Study Hours')
plt.ylabel('Final Exam Marks')
plt.title('Study Hours vs Final Exam Marks')
plt.show()
```

```python
plt.scatter(df['Attendance_%'], df['Final_Exam_Marks'])
plt.xlabel('Attendance %')
plt.ylabel('Final Exam Marks')
plt.title('Attendance vs Final Exam Marks')
plt.show()
```

## Correlation Analysis

```python
corr = df_num.corr()
corr
```

```python
plt.matshow(corr, cmap='coolwarm')
plt.xticks(range(len(num_cols)), num_cols, rotation=90)
plt.yticks(range(len(num_cols)), num_cols)
plt.colorbar()
plt.title('Correlation Matrix', pad=60)
plt.show()
```

## Observations

- Internal_Marks has the strongest positive correlation with Final_Exam_Marks (about 0.60), which is a moderate relationship.
- Assignment_Marks (about 0.39) and Study_Hours (about 0.37) also show weak-to-moderate positive correlation with Final_Exam_Marks.
- Attendance_% has the weakest positive correlation with Final_Exam_Marks (about 0.32) among the four variables.
- Final_Exam_Marks ranges from 49 to 96, with a mean of about 71.1 and a standard deviation of about 10.5.
- All correlations are positive, meaning higher values of these variables are generally associated with higher final exam marks, but none of these relationships are very strong, so other factors are also affecting performance.

## Conclusion

The dataset of 100 students was clean, with no missing values or duplicates. All five academic variables are numerical, and there are no categorical variables in this dataset. Among the variables studied, Internal_Marks showed the highest positive correlation with Final_Exam_Marks, while Attendance_% showed the lowest. Overall, the analysis shows a positive but moderate relationship between the given academic factors and final exam performance.
