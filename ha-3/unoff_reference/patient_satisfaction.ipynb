To find out whether patient satisfaction differs significantly (a) between three hospitals (A, B, C) and (b) between four age groups (18-30, 31-45, 46-60, 60+), and to support the conclusion with confidence intervals, hypothesis tests and one-way ANOVA.

## 3. Sampling and sampling distribution
- A simple random sample (n = 40) has mean 3.963, and a proportionate stratified sample by hospital (n = 40) has mean 4.035. Both are close to the full-survey mean of 3.976.
- Drawing 2000 random samples of size 30 gives sample means averaging 3.973 with standard deviation 0.084. The theoretical standard error (sigma / sqrt(n)) is 0.094.
- The histogram of sample means is bell-shaped, as the **Central Limit Theorem** predicts. Normal-theory confidence intervals and tests are therefore justified.

## 4. Estimation: 95% confidence intervals (t-interval)
The population standard deviation is unknown, so the t-interval x-bar +/- t(0.025, n-1) x s / sqrt(n) is used.

## 5. Hypothesis tests

### 5.1 One-sample tests against a benchmark of 4.0
H0: mu = 4.0  vs  H1: mu != 4.0 (two-tailed). Because n = 120 is large, the Z-test uses the sample standard deviation in place of sigma.

| Test | Statistic | Critical value | p-value | Decision |
|---|---|---|---|---|
| One-sample Z-test | -0.511 | +/- 1.960 | 0.609 | Fail to reject H0 |
| One-sample t-test (df = 119) | -0.511 | +/- 1.980 | 0.610 | Fail to reject H0 |

Overall satisfaction is not significantly different from 4.0.

### 5.2 Independent two-sample t-tests
Levene's test supported equal variances in both cases, so the pooled t-test was used.

| Comparison | Mean difference | t | p-value | Decision |
|---|---|---|---|---|
| Hospital B vs C | -0.369 | -3.334 | **0.0013** | Reject H0 |
| Age 31-45 vs 18-30 | +0.164 | 1.241 | 0.220 | Fail to reject H0 |

### 5.3 Paired t-test
**Not applicable.** A paired t-test needs two measurements on the same patients. Here each patient has one score and belongs to one hospital and one age group, so the groups are independent.

## 6. One-way ANOVA
H0: all group means are equal.  H1: at least one group mean differs.

**Assumption checks**

| Factor | Shapiro-Wilk (normality) p-values | Levene (equal variance) p |
|---|---|---|
| Hospital | A 0.797, B 0.647, C 0.171 | 0.760 |
| Age group | 0.584, 0.764, 0.418, 0.251 | 0.617 |

All p-values exceed 0.05, so the assumptions hold.

**ANOVA: hospitals**

| Source | Sum of squares | df | Mean square | F | p-value |
|---|---|---|---|---|---|
| Between hospitals | 3.005 | 2 | 1.503 | 6.076 | **0.0031** |
| Within (error) | 28.935 | 117 | 0.247 | | |
| Total | 31.940 | 119 | | | |

Eta-squared = 0.094. **Decision: reject H0.**

**ANOVA: age groups**

| Source | Sum of squares | df | Mean square | F | p-value |
|---|---|---|---|---|---|
| Between age groups | 0.404 | 3 | 0.135 | 0.495 | 0.687 |
| Within (error) | 31.536 | 116 | 0.272 | | |
| Total | 31.940 | 119 | | | |

Eta-squared = 0.013. **Decision: fail to reject H0.**

**Follow-up pairwise comparisons for hospitals** (independent t-tests, Bonferroni level 0.05 / 3 = 0.0167)

| Pair | Mean difference | t | p-value | Significant? |
|---|---|---|---|---|
| A vs B | +0.243 | 2.081 | 0.041 | No |
| A vs C | -0.126 | -1.180 | 0.242 | No |
| B vs C | -0.369 | -3.334 | 0.0013 | **Yes** |

## 7. Conclusions
1. **Hospital matters.** Satisfaction differs significantly between hospitals (ANOVA F = 6.08, p = 0.0031). The difference lies between Hospital B (lowest, mean 3.77) and Hospital C (highest, mean 4.14).
2. **Age group does not matter.** There is no evidence of a difference across age groups (F = 0.49, p = 0.687). The observed differences are consistent with random variation.
3. **Overall satisfaction** (3.98, 95% CI 3.88 to 4.07) is not significantly different from the 4.0 benchmark.
4. **Effect size.** Hospital explains about 9.4% of the variation in satisfaction, a modest effect. Most of the variation is between individual patients.
5. **Recommendation.** Review service quality at Hospital B. The survey is observational, so the analysis shows that a difference exists but not why.

## 8. Limitations
- The sample is 120 patients, and group sizes are uneven (26 to 44).
- The 4.0 benchmark is an assumed target.
- Hospital and age group are not evenly mixed (for example, Hospital B has 17 patients aged 18-30 against 6 in Hospital C), so a hospital effect could partly reflect patient mix.

## 1. Import libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

## 2. Load the data and clean it

```python
df = pd.read_csv("patient_survey.csv")

print(df.head())
print("Rows and columns:", df.shape)
print()
print(df.dtypes)
print()
print("Missing values in each column:")
print(df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())
print()
print(df["Hospital"].value_counts())
print(df["Age_Group"].value_counts())
```

No missing values and no duplicates, so no cleaning is needed.

**Types of data:** Hospital is nominal (categorical), Age_Group is ordinal (categorical), Satisfaction_Score is numerical (continuous).

## 3. Exploratory Data Analysis (EDA)

```python
score = df["Satisfaction_Score"]

print("Mean   :", round(score.mean(), 3))
print("Median :", score.median())
print("Mode   :", score.mode()[0])
print("Std dev:", round(score.std(), 3))
print("Min, Max:", score.min(), score.max())
```

```python
# mean, median and spread for each hospital and each age group
print("By hospital:")
print(df.groupby("Hospital")["Satisfaction_Score"].agg(["count", "mean", "median", "std"]).round(3))
print()
print("By age group:")
print(df.groupby("Age_Group")["Satisfaction_Score"].agg(["count", "mean", "median", "std"]).round(3))
```

## 4. Sampling techniques
We select two samples from the 120 patients: a simple random sample and a stratified sample (equal number from each hospital).

```python
# simple random sample of 40 patients
srs = df.sample(n=40, random_state=1)

# stratified sample: 15 patients from each hospital
stratified = df.groupby("Hospital").sample(n=15, random_state=1)

print("Mean of all 120 patients :", round(score.mean(), 3))
print("Mean of random sample    :", round(srs["Satisfaction_Score"].mean(), 3))
print("Mean of stratified sample:", round(stratified["Satisfaction_Score"].mean(), 3))
```

## 5. Sampling distribution and Central Limit Theorem
We take 1000 random samples of size 30 and look at the distribution of the sample means.

```python
np.random.seed(1)

sample_means = []
for i in range(1000):
    sample = df["Satisfaction_Score"].sample(n=30, replace=True)
    sample_means.append(sample.mean())

print("Population mean            :", round(score.mean(), 3))
print("Mean of the sample means   :", round(np.mean(sample_means), 3))
print("Std of the sample means    :", round(np.std(sample_means), 3))
print("Standard error s/sqrt(n)   :", round(score.std() / np.sqrt(30), 3))

plt.hist(sample_means, bins=25, color="skyblue", edgecolor="black")
plt.xlabel("Sample mean")
plt.ylabel("Frequency")
plt.title("Sampling distribution of the mean (n = 30)")
plt.show()
```

The sample means form a bell shape around the population mean, and their standard deviation is close to s/sqrt(n). This is the Central Limit Theorem.

## 6. Estimation: 95% confidence intervals
The population standard deviation is unknown, so we use the t distribution: CI = mean ± t × s / sqrt(n).

```python
def confidence_interval(data):
    n = len(data)
    mean = data.mean()
    se = data.std() / np.sqrt(n)               # standard error
    t_value = stats.t.ppf(0.975, n - 1)        # 95% confidence
    return mean - t_value * se, mean + t_value * se

# whole data and the samples
for name, data in [("All 120 patients", score),
                   ("Random sample (40)", srs["Satisfaction_Score"]),
                   ("Stratified sample (45)", stratified["Satisfaction_Score"])]:
    low, high = confidence_interval(data)
    print(name, ": mean =", round(data.mean(), 3), ", 95% CI = (", round(low, 3), ",", round(high, 3), ")")
```

```python
print("95% CI for each hospital:")
for h in ["A", "B", "C"]:
    data = df[df["Hospital"] == h]["Satisfaction_Score"]
    low, high = confidence_interval(data)
    print("Hospital", h, ": mean =", round(data.mean(), 3), ", CI = (", round(low, 3), ",", round(high, 3), ")")

print()
print("95% CI for each age group:")
for a in ["18-30", "31-45", "46-60", "60+"]:
    data = df[df["Age_Group"] == a]["Satisfaction_Score"]
    low, high = confidence_interval(data)
    print("Age", a, ": mean =", round(data.mean(), 3), ", CI = (", round(low, 3), ",", round(high, 3), ")")
```

## 7. Hypothesis formulation
The hospital expects an average satisfaction of 4.0.

- H0: population mean = 4.0
- H1: population mean ≠ 4.0 (two-tailed)
- Significance level: α = 0.05. If p-value < 0.05 we reject H0.

## 8. One-sample Z-test
The sample is large (n = 120), so the sample standard deviation is used in place of the population standard deviation.

```python
mu0 = 4.0
n = len(score)

z = (score.mean() - mu0) / (score.std() / np.sqrt(n))
p_value = 2 * (1 - stats.norm.cdf(abs(z)))        # two-tailed

print("Z statistic:", round(z, 3))
print("p-value    :", round(p_value, 4))
if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

## 9. One-sample t-test
Same hypothesis for the whole data, then for each hospital separately.

```python
t_stat, p_value = stats.ttest_1samp(score, 4.0)
print("All patients: t =", round(t_stat, 3), ", p-value =", round(p_value, 4))

for h in ["A", "B", "C"]:
    data = df[df["Hospital"] == h]["Satisfaction_Score"]
    t_stat, p_value = stats.ttest_1samp(data, 4.0)
    print("Hospital", h, ": mean =", round(data.mean(), 3), ", t =", round(t_stat, 3), ", p-value =", round(p_value, 4))
```

## 10. Independent t-test
Compare the average satisfaction of two hospitals at a time.

- H0: the two hospitals have the same mean satisfaction
- H1: the means are different

```python
hosp_A = df[df["Hospital"] == "A"]["Satisfaction_Score"]
hosp_B = df[df["Hospital"] == "B"]["Satisfaction_Score"]
hosp_C = df[df["Hospital"] == "C"]["Satisfaction_Score"]

t1, p1 = stats.ttest_ind(hosp_A, hosp_B)
t2, p2 = stats.ttest_ind(hosp_A, hosp_C)
t3, p3 = stats.ttest_ind(hosp_B, hosp_C)

print("A vs B: t =", round(t1, 3), ", p-value =", round(p1, 4))
print("A vs C: t =", round(t2, 3), ", p-value =", round(p2, 4))
print("B vs C: t =", round(t3, 3), ", p-value =", round(p3, 4))
```

A paired t-test is not used here because this survey has no paired (before and after) measurements on the same patients.

## 11. One-way ANOVA
ANOVA compares the means of more than two groups at once.

**Hospital:** H0: mean satisfaction is the same for A, B and C. H1: at least one hospital has a different mean.

**Age group:** H0: mean satisfaction is the same for all four age groups. H1: at least one age group has a different mean.

(The group standard deviations in section 3 are similar, which is the equal-variance condition for ANOVA.)

```python
# ANOVA for hospitals
F1, p_hosp = stats.f_oneway(hosp_A, hosp_B, hosp_C)
print("Hospital  : F =", round(F1, 3), ", p-value =", round(p_hosp, 4))

# ANOVA for age groups
age1 = df[df["Age_Group"] == "18-30"]["Satisfaction_Score"]
age2 = df[df["Age_Group"] == "31-45"]["Satisfaction_Score"]
age3 = df[df["Age_Group"] == "46-60"]["Satisfaction_Score"]
age4 = df[df["Age_Group"] == "60+"]["Satisfaction_Score"]

F2, p_age = stats.f_oneway(age1, age2, age3, age4)
print("Age group : F =", round(F2, 3), ", p-value =", round(p_age, 4))

print()
if p_hosp < 0.05:
    print("Hospital: reject H0 (significant difference)")
else:
    print("Hospital: fail to reject H0")
if p_age < 0.05:
    print("Age group: reject H0 (significant difference)")
else:
    print("Age group: fail to reject H0")
```

## 12. Visualization

```python
plt.figure(figsize=(14, 5))

# boxplot by hospital
plt.subplot(1, 3, 1)
plt.boxplot([hosp_A, hosp_B, hosp_C], tick_labels=["A", "B", "C"])
plt.xlabel("Hospital")
plt.ylabel("Satisfaction score")
plt.title("Satisfaction by hospital")

# boxplot by age group
plt.subplot(1, 3, 2)
plt.boxplot([age1, age2, age3, age4], tick_labels=["18-30", "31-45", "46-60", "60+"])
plt.xlabel("Age group")
plt.title("Satisfaction by age group")

# hospital means with 95% confidence intervals
plt.subplot(1, 3, 3)
means = []
errors = []
for data in [hosp_A, hosp_B, hosp_C]:
    low, high = confidence_interval(data)
    means.append(data.mean())
    errors.append((high - low) / 2)
plt.errorbar(["A", "B", "C"], means, yerr=errors, fmt="o", capsize=6)
plt.axhline(4.0, color="red", linestyle="--", label="Benchmark 4.0")
plt.xlabel("Hospital")
plt.ylabel("Mean satisfaction")
plt.title("Mean with 95% CI")
plt.legend()

plt.tight_layout()
plt.show()
```
