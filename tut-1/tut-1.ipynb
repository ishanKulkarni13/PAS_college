```python
import pandas as pd
import numpy as np
import time
from tabulate import tabulate

print("Welcome to the most advanced ")

# Manual 

def mean(A):
    total = 0

    for x in A:
        total += x

    return total / len(A)


def median(A):
    n = len(A)

    if n % 2 == 1:
        return A[n // 2]

    return (A[n // 2 - 1] + A[n // 2]) / 2


def mode(A):
    mode = A[0]
    max_count = 0

    for i in range(len(A)):
        count = 0

        for j in range(len(A)):
            if A[i] == A[j]:
                count += 1

        if count > max_count:
            max_count = count
            mode = A[i]

    return mode


def variance(A):
    m = mean(A)

    total = 0

    for x in A:
        total += (x - m) ** 2

    return total / len(A)


def std_dev(A):
    return variance(A) ** 0.5


def quartiles(A):
    n = len(A)

    if n % 2 == 0:
        lower = A[:n // 2]
        upper = A[n // 2:]
    else:
        lower = A[:n // 2]
        upper = A[n // 2 + 1:]

    q0 = A[0]
    q1 = median(lower)
    q2 = median(A)
    q3 = median(upper)
    q4 = A[-1]

    return q0, q1, q2, q3, q4


def iqr(A):
    q0, q1, q2, q3, q4 = quartiles(A)
    return q3 - q1


def range_value(A):
    maximum = A[0]
    minimum = A[0]

    for x in A:

        if x > maximum:
            maximum = x

        if x < minimum:
            minimum = x

    return maximum - minimum



# Read tha dataset

# df = pd.read_csv("./data/Tutorial_1_Datasets_Probability_Statistics.csv")
df = pd.read_csv("./data/test.csv")
df = df.dropna()

subjects = ["Sub1", "Sub2", "Sub3", "Sub4"]

for subject in subjects:

    print("\n")
    print("=" * 70)
    print(subject)
    print("=" * 70)

    A = df[subject].astype(float).tolist()
    A.sort()

    # Manual

    start = time.perf_counter()

    manual_mean = mean(A)
    manual_median = median(A)
    manual_mode = mode(A)
    manual_variance = variance(A)
    manual_std = std_dev(A)

    q0, q1, q2, q3, q4 = quartiles(A)

    manual_iqr = iqr(A)
    manual_range = range_value(A)

    manual_time = (time.perf_counter() - start) * 1000

    # NumPy Pandas

    start = time.perf_counter()

    np_mean = np.mean(A)
    np_median = np.median(A)

    mode_series = pd.Series(A).mode()
    np_mode = mode_series.iloc[0]

    np_variance = np.var(A)
    np_std = np.std(A)

    np_q0 = np.min(A)
    np_q1 = np.percentile(A, 25)
    np_q2 = np.percentile(A, 50)
    np_q3 = np.percentile(A, 75)
    np_q4 = np.max(A)

    np_iqr = np_q3 - np_q1
    np_range = np.max(A) - np.min(A)

    numpy_time = (time.perf_counter() - start) * 1000
    # display tha output

    table = [
    ["Count", len(A), len(A)],
    ["Mean", manual_mean, np_mean],
    ["Median", manual_median, np_median],
    ["Mode", manual_mode, np_mode],
    ["Variance", manual_variance, np_variance],
    ["Std Dev", manual_std, np_std],
    ["Q0", q0, np_q0],
    ["Q1", q1, np_q1],
    ["Q2", q2, np_q2],
    ["Q3", q3, np_q3],
    ["Q4", q4, np_q4],
    ["IQR", manual_iqr, np_iqr],
    ["Range", manual_range, np_range],
]

print(f"\n{subject}")

print(tabulate(
    table,
    headers=["Statistic", "Manual", "NumPy/Pandas"],
    tablefmt="grid",
    floatfmt=".4f"
))

print(f"\nManual Time : {manual_time:.4f} ms")
print(f"NumPy Time  : {numpy_time:.4f} ms")
```
