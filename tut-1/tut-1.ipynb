```python
import pandas as pd
import numpy as np
import time
from tabulate import tabulate
# FILE_PATH = "./data/Tutorial_1_Datasets_Probability_Statistics.csv"
FILE_PATH = "./data/test.csv"

print("Welcome to the most advanced program")


# Manual Functions

def mean(A):
    total = 0

    for x in A:
        total += x

    return total / len(A)


def median(A):
    n = len(A)

    if n % 2 == 1:
        return A[n // 2]

    return (A[n // 2 - 1] + A[n // 2]) / 2



def mode(A): #n2 
    mode = A[0]
    max_count = 0

    for i in range(len(A)):
        count = 0

        for j in range(len(A)):
            if A[i] == A[j]:
                count += 1

        if count > max_count:
            max_count = count
            mode = A[i]

    return mode

# def mode(A): #optimized and blessed by me

#     freq = {}

#     # Count frequency of each value
#     for x in A:
#         if x in freq:
#             freq[x] += 1
#         else:
#             freq[x] = 1

#     # Find the value with maximum frequency
#     mode = None
#     max_count = 0

#     for key in freq:
#         if freq[key] > max_count:
#             max_count = freq[key]
#             mode = key

#     return mode

def variance(A):
    m = mean(A)

    total = 0

    for x in A:
        total += (x - m) ** 2

    return total / len(A)


def std_dev(A):
    return variance(A) ** 0.5


def quartiles(A):
    n = len(A)

    if n % 2 == 0:
        lower = A[:n // 2]
        upper = A[n // 2:]
    else:
        lower = A[:n // 2]
        upper = A[n // 2 + 1:]

    q0 = A[0]
    q1 = median(lower)
    q2 = median(A)
    q3 = median(upper)
    q4 = A[-1]

    return q0, q1, q2, q3, q4


def range_value(A):
    maximum = A[0]
    minimum = A[0]

    for x in A:

        if x > maximum:
            maximum = x

        if x < minimum:
            minimum = x

    return maximum - minimum


  
# Read Dataset
  

# dataset = pd.read_csv("./data/test.csv") 
#dataset = pd.read_csv("./data/Tutorial_1_Datasets_Probability_Statistics.csv")

dataset = pd.read_csv(FILE_PATH)


dataset = dataset.dropna()

subjects = ["Sub1", "Sub2", "Sub3", "Sub4"]


  
# Store Results
  

results = {}

statistics = [
    "Count",
    "Mean",
    "Median",
    "Mode",
    "Variance",
    "Std Dev",
    "Q0",
    "Q1",
    "Q2",
    "Q3",
    "Q4",
    "IQR",
    "Range"
]


# Time for each statistic
manual_times = [0.0] * len(statistics)
numpy_times = [0.0] * len(statistics)


  
# Process Each Subject
  

for subject in subjects:

    A = dataset[subject].astype(float).tolist()

    A.sort()


     
    # Manual Mean
     

    start = time.perf_counter()

    manual_mean = mean(A)

    manual_times[1] += (time.perf_counter() - start) * 1000


     
    # Manual Median
     

    start = time.perf_counter()

    manual_median = median(A)

    manual_times[2] += (time.perf_counter() - start) * 1000


     
    # Manual Mode
     

    start = time.perf_counter()

    manual_mode = mode(A)

    manual_times[3] += (time.perf_counter() - start) * 1000


    # Manual Variance

    start = time.perf_counter()

    manual_variance = variance(A)

    manual_times[4] += (time.perf_counter() - start) * 1000


    # Manual Standard Deviation

    start = time.perf_counter()

    manual_std = manual_variance ** 0.5

    manual_times[5] += (time.perf_counter() - start) * 1000


    # Manual Quartiles

    start = time.perf_counter()

    q0, q1, q2, q3, q4 = quartiles(A)

    quartile_time = (time.perf_counter() - start) * 1000


    # Store quartile time for Q1, Q2 and Q3
    manual_times[7] += quartile_time / 3
    manual_times[8] += quartile_time / 3
    manual_times[9] += quartile_time / 3


    # Q0 and Q4 are just first and last elements
    start = time.perf_counter()

    manual_q0 = A[0]

    manual_times[6] += (time.perf_counter() - start) * 1000


    start = time.perf_counter()

    manual_q4 = A[-1]

    manual_times[10] += (time.perf_counter() - start) * 1000


    # Manual IQR

    start = time.perf_counter()

    manual_iqr = q3 - q1

    manual_times[11] += (time.perf_counter() - start) * 1000


    # Manual Range

    start = time.perf_counter()

    manual_range = range_value(A)

    manual_times[12] += (time.perf_counter() - start) * 1000



    # Mean
    start = time.perf_counter()

    np_mean = np.mean(A)

    numpy_times[1] += (time.perf_counter() - start) * 1000


    # Median
    start = time.perf_counter()

    np_median = np.median(A)

    numpy_times[2] += (time.perf_counter() - start) * 1000


    # Mode
    start = time.perf_counter()

    np_mode = pd.Series(A).mode().iloc[0]

    numpy_times[3] += (time.perf_counter() - start) * 1000


    # Variance
    start = time.perf_counter()

    np_variance = np.var(A)

    numpy_times[4] += (time.perf_counter() - start) * 1000


    # Standard Deviation
    start = time.perf_counter()

    np_std = np.std(A)

    numpy_times[5] += (time.perf_counter() - start) * 1000


    # Q0
    start = time.perf_counter()

    np_q0 = np.min(A)

    numpy_times[6] += (time.perf_counter() - start) * 1000


    # Q1
    start = time.perf_counter()

    np_q1 = np.percentile(A, 25)

    numpy_times[7] += (time.perf_counter() - start) * 1000


    # Q2
    start = time.perf_counter()

    np_q2 = np.percentile(A, 50)

    numpy_times[8] += (time.perf_counter() - start) * 1000


    # Q3
    start = time.perf_counter()

    np_q3 = np.percentile(A, 75)

    numpy_times[9] += (time.perf_counter() - start) * 1000


    # Q4
    start = time.perf_counter()

    np_q4 = np.max(A)

    numpy_times[10] += (time.perf_counter() - start) * 1000


    # IQR
    start = time.perf_counter()

    np_iqr = np_q3 - np_q1

    numpy_times[11] += (time.perf_counter() - start) * 1000


    # Range
    start = time.perf_counter()

    np_range = np.max(A) - np.min(A)

    numpy_times[12] += (time.perf_counter() - start) * 1000


    # Store manual results

    results[subject] = [
        len(A),
        manual_mean,
        manual_median,
        manual_mode,
        manual_variance,
        manual_std,
        manual_q0,
        q1,
        q2,
        q3,
        manual_q4,
        manual_iqr,
        manual_range
    ]


# Create final table

table = []

for i in range(len(statistics)):

    row = [
        statistics[i],
        results["Sub1"][i],
        results["Sub2"][i],
        results["Sub3"][i],
        results["Sub4"][i]
    ]

    if i == 0:
        row.append("-")
        row.append("-")
    else:
        row.append(manual_times[i])
        row.append(numpy_times[i])

    table.append(row)


# Display table

print()

print(
    tabulate(
        table,
        headers=[
            "Statistic",
            "Sub1",
            "Sub2",
            "Sub3",
            "Sub4",
            "Total Manual Time (ms)",
            "Total NumPy Time (ms)"
        ],
        tablefmt="grid",
        floatfmt=".4f"
    )
)
```
