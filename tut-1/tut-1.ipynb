```python
import pandas as pd
import numpy as np
import time
from tabulate import tabulate

print("Welcome to the most advanced")

# =============================================================================
# Manual Functions
# =============================================================================

def mean(A):
    total = 0
    for x in A:
        total += x
    return total / len(A)


def median(A):
    n = len(A)

    if n % 2 == 1:
        return A[n // 2]

    return (A[n // 2 - 1] + A[n // 2]) / 2


def mode(A):
    mode = A[0]
    max_count = 0

    for i in range(len(A)):
        count = 0

        for j in range(len(A)):
            if A[i] == A[j]:
                count += 1

        if count > max_count:
            max_count = count
            mode = A[i]

    return mode


def variance(A):
    m = mean(A)

    total = 0

    for x in A:
        total += (x - m) ** 2

    return total / len(A)


def std_dev(A):
    return variance(A) ** 0.5


def quartiles(A):
    n = len(A)

    if n % 2 == 0:
        lower = A[:n // 2]
        upper = A[n // 2:]
    else:
        lower = A[:n // 2]
        upper = A[n // 2 + 1:]

    q0 = A[0]
    q1 = median(lower)
    q2 = median(A)
    q3 = median(upper)
    q4 = A[-1]

    return q0, q1, q2, q3, q4


def iqr(A):
    q0, q1, q2, q3, q4 = quartiles(A)
    return q3 - q1


def range_value(A):
    maximum = A[0]
    minimum = A[0]

    for x in A:
        if x > maximum:
            maximum = x

        if x < minimum:
            minimum = x

    return maximum - minimum


# =============================================================================
# Read Dataset
# =============================================================================

# df = pd.read_csv("./data/Tutorial_1_Datasets_Probability_Statistics.csv")
df = pd.read_csv("./data/test.csv")
df = df.dropna()

subjects = ["Sub1", "Sub2", "Sub3", "Sub4"]

results = {}

total_manual_time = 0
total_numpy_time = 0

# =============================================================================
# Process Each Subject
# =============================================================================

for subject in subjects:

    A = df[subject].astype(float).tolist()
    A.sort()

    # -------------------------------------------------------------------------
    # Manual
    # -------------------------------------------------------------------------

    start = time.perf_counter()

    manual_mean = mean(A)
    manual_median = median(A)
    manual_mode = mode(A)
    manual_variance = variance(A)
    manual_std = std_dev(A)

    q0, q1, q2, q3, q4 = quartiles(A)

    manual_iqr = iqr(A)
    manual_range = range_value(A)

    manual_time = (time.perf_counter() - start) * 1000

    total_manual_time += manual_time

    # -------------------------------------------------------------------------
    # NumPy / Pandas
    # -------------------------------------------------------------------------

    start = time.perf_counter()

    np.mean(A)
    np.median(A)
    pd.Series(A).mode().iloc[0]
    np.var(A)
    np.std(A)
    np.min(A)
    np.percentile(A, 25)
    np.percentile(A, 50)
    np.percentile(A, 75)
    np.max(A)

    numpy_time = (time.perf_counter() - start) * 1000

    total_numpy_time += numpy_time

    # -------------------------------------------------------------------------
    # Store Results
    # -------------------------------------------------------------------------

    results[subject] = {
        "Count": len(A),
        "Mean": manual_mean,
        "Median": manual_median,
        "Mode": manual_mode,
        "Variance": manual_variance,
        "Std Dev": manual_std,
        "Q0": q0,
        "Q1": q1,
        "Q2": q2,
        "Q3": q3,
        "Q4": q4,
        "IQR": manual_iqr,
        "Range": manual_range,
    }

# =============================================================================
# Final Summary Table
# =============================================================================

statistics = [
    "Count",
    "Mean",
    "Median",
    "Mode",
    "Variance",
    "Std Dev",
    "Q0",
    "Q1",
    "Q2",
    "Q3",
    "Q4",
    "IQR",
    "Range",
]

table = []

for stat in statistics: # building tha rows just like ironman, in a cave, with bunch of scraps for emotional support

    row = [
        stat,
        results["Sub1"][stat],
        results["Sub2"][stat],
        results["Sub3"][stat],
        results["Sub4"][stat],
    ]

    if stat == "Count":
        row.append("-")
        row.append("-")
    else:
        row.append(total_manual_time)
        row.append(total_numpy_time)

    table.append(row)


print(
    tabulate(
        table,
        headers=[
            "Statistic",
            "Sub1",
            "Sub2",
            "Sub3",
            "Sub4",
            "Total Manual Time (ms)",
            "Total NumPy Time (ms)",
        ],
        tablefmt="grid",
        floatfmt=".4f",
    )
)
```
