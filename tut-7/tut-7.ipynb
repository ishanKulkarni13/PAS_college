```python
import math
import numpy as np
from scipy.stats import norm
from scipy.stats import ttest_ind, ttest_rel
```

```python
mu0 = float(input("Enter hypothesized population mean: "))
x_bar = float(input("Enter sample mean: "))
n = int(input("Enter sample size: "))
sigma = float(input("Enter population standard deviation: "))
alpha = float(input("Enter significance level: "))
```

```python
print("\nSelect Test:")
print("1. mu is less than mu0")
print("2. mu is greater than mu0")
print("3. mu is not equal to mu0")

option = int(input("Enter option (1/2/3): "))
```

```python
z = (x_bar - mu0) / (sigma / math.sqrt(n))

print("\nZ TEST RESULT")
print("Hypothesized Mean :", mu0)
print("Sample Mean :", x_bar)
print("Sample Size :", n)
print("Population Std :", sigma)
print("Significance Level :", alpha)
print("Calculated Z :", round(z, 4))
```

```python
if option == 1:

    p_value = norm.cdf(z)

    print("\nLeft-Tailed Test")
    print("H₀ : μ = μ₀")
    print("H₁ : μ < μ₀")
    print("P-value :", round(p_value, 4))

    if p_value < alpha:
        print("Decision : Reject H₀")
    else:
        print("Decision : Fail to Reject H₀")


elif option == 2:

    p_value = 1 - norm.cdf(z)

    print("\nRight-Tailed Test")
    print("H₀ : μ = μ₀")
    print("H₁ : μ > μ₀")
    print("P-value :", round(p_value, 4))

    if p_value < alpha:
        print("Decision : Reject H₀")
    else:
        print("Decision : Fail to Reject H₀")


elif option == 3:

    p_value = 2 * (1 - norm.cdf(abs(z)))

    print("\nTwo-Tailed Test")
    print("H₀ : μ = μ₀")
    print("H₁ : μ ≠ μ₀")
    print("P-value :", round(p_value, 4))

    if p_value < alpha:
        print("Decision : Reject H₀")
    else:
        print("Decision : Fail to Reject H₀")


else:

    print("Invalid option")
```

```python
group_A = [72, 75, 68, 70, 74, 69, 77, 73, 71, 76]

group_B = [65, 67, 70, 64, 68, 66, 69, 63, 67, 65]

print("Group A :", group_A)
print("Group B :", group_B)
```

```python
t_stat, p_value = ttest_ind(group_A, group_B)

print("\nINDEPENDENT TWO-SAMPLE t-TEST")

print("Group A Mean :", round(np.mean(group_A), 2))
print("Group B Mean :", round(np.mean(group_B), 2))
print("T-statistic :", round(t_stat, 4))
print("P-value :", round(p_value, 4))
```

```python
alpha = 0.05

print("\nH₀ : Mean of Group A = Mean of Group B")
print("H₁ : Mean of Group A ≠ Mean of Group B")

if p_value < alpha:
    print("Decision : Reject H₀")
    print("Conclusion : There is a significant difference between the two groups.")
else:
    print("Decision : Fail to Reject H₀")
    print("Conclusion : There is no significant difference between the two groups.")
```

```python
before = [60, 65, 58, 70, 62, 67, 64, 61, 59, 63]

after = [68, 72, 65, 75, 69, 73, 70, 67, 66, 71]

print("Before :", before)
print("After  :", after)
```

```python
t_stat, p_value = ttest_rel(before, after)

print("\nPAIRED t-TEST")

print("Before Mean :", round(np.mean(before), 2))
print("After Mean :", round(np.mean(after), 2))
print("T-statistic :", round(t_stat, 4))
print("P-value :", round(p_value, 4))
```

```python
alpha = 0.05

print("\nH₀ : There is no significant difference")
print("H₁ : There is a significant difference")

if p_value < alpha:
    print("Decision : Reject H₀")
    print("Conclusion : There is a significant difference between before and after scores.")
else:
    print("Decision : Fail to Reject H₀")
    print("Conclusion : There is no significant difference between before and after scores.")
```

```python
print("COMPARISON OF TESTS")

print("\n1. One-Sample Z-Test")
print("Used to compare a sample mean with a population mean.")

print("\n2. Independent Two-Sample t-Test")
print("Used to compare the means of two independent groups.")

print("\n3. Paired t-Test")
print("Used to compare two related measurements.")

print("\nP-value Rule:")
print("If p-value < 0.05  -> Reject H₀")
print("If p-value >= 0.05 -> Fail to Reject H₀")
```
