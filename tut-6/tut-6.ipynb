```python
import math
from scipy.stats import norm
```

```python
mu0 = 50
x_bar = 52
n = 100
sigma = 10
alpha = 0.05

# 1 = left, 2 = right, 3 = two tailed
option = 3
```

```python
def find_z(x_bar, mu0, sigma, n):
    return (x_bar - mu0) / (sigma / math.sqrt(n))

z = find_z(x_bar, mu0, sigma, n)
```

```python
print("z test result: ")
print("mu0 =", mu0)
print("sample mean =", x_bar)
print("n =", n)
print("sigma =", sigma)
print("alpha =", alpha)
print("z =", round(z, 4))
```

```python
def left_tail(z, alpha):
    z_critical = norm.ppf(alpha)
    print("left tailed test")
    print("h0: mu = mu0")
    print("h1: mu < mu0")
    print("critical z =", round(z_critical, 4))

    if z < z_critical:
        print("reject h0")
    else:
        print("fail to reject h0")


def right_tail(z, alpha):
    z_critical = norm.ppf(1 - alpha)
    print("right tailed test")
    print("h0: mu = mu0")
    print("h1: mu > mu0")
    print("critical z =", round(z_critical, 4))

    if z > z_critical:
        print("reject h0")
    else:
        print("fail to reject h0")


def two_tail(z, alpha):
    z_lower = norm.ppf(alpha / 2)
    z_upper = norm.ppf(1 - alpha / 2)
    print("two tailed test")
    print("h0: mu = mu0")
    print("h1: mu != mu0")
    print("lower critical z =", round(z_lower, 4))
    print("upper critical z =", round(z_upper, 4))

    if z < z_lower or z > z_upper:
        print("reject h0")
    else:
        print("fail to reject h0")
```

```python
if option == 1:
    left_tail(z, alpha)
elif option == 2:
    right_tail(z, alpha)
elif option == 3:
    two_tail(z, alpha)
else:
    print("nigga, wrong option")
```
