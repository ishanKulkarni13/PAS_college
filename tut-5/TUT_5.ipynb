```python
import numpy as np
```

```python
P = np.random.exponential(5,500000)
```

```python
P
```

```python
N = 40
n = 40000
sample_mean = []
```

```python
for i in range(N):
    sample = np.random.choice(P,n,replace = False)
    sample_mean.append(np.mean(sample))
```

```python
print("P - mean : ",np.mean(P))
print("S - mean : ",np.mean(sample_mean))
```

```python
P_std = np.std(P)
```

```python
print("P-std : ",P_std)
print("S-std : ",np.std(sample_mean))
print("S_P std : ",P_std/(n**0.5))
```

```python
import scipy.stats as st
```

```python
CL = np.arange(80, 100)
```

```python
CL
```

```python
X = np.mean(sample_mean)
S_std = P_std / (n**0.5)
```

```python
print(f"{'Confidence':<15} | {'Alpha':<10} | {'Z-Score':<10} | {'CI Min':<15} | {'CI Max':<15}")
print("-" * 75)

for c in CL:
   
    alpha = 1 - (c / 100)
    z_score = st.norm.ppf(1 - alpha/2)
    
    error = z_score * S_std
    mini = X - error
    maxi = X + error
    
    print(f"{c:<15} | {alpha:<10.2f} | {z_score:<10.3f} | {mini:<15.4f} | {maxi:<15.4f}")
    print("-" * 75)
```
